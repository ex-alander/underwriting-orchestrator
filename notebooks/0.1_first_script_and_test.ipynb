## 1. Функция, которую будем тестировать

Пусть у нас есть функция, которая считает **DTI** — debt-to-income ratio. Это одна из ключевых метрик в кредитном андеррайтинге: отношение ежемесячного долга к ежемесячному доходу.

Формула:

```
DTI = monthly_debt / monthly_income
```

Например, если клиент платит 30 000 ₽ в месяц по кредитам при доходе 100 000 ₽, то DTI = 0.3 (или 30%).

Обычно банки отсекают заявки с DTI > 0.5. Это одно из правил, которое мы потом проверим в policy check агенте.

````python
def calc_dti(monthly_debt: float, monthly_income: float) -> float:
    """Возвращает debt-to-income ratio."""
    if monthly_income <= 0:
        raise ValueError("monthly_income must be positive")
    return monthly_debt / monthly_income
````

## 2. Пишем тест

Тест — это функция, которая **проверяет утверждение о поведении кода**. Если утверждение не выполняется, тест падает с ошибкой.

Простейший способ в Python — использовать библиотеку `pytest`. Но чтобы не ставить её сразу, начнём с обычных `assert` — их можно писать прямо в тетради.

**Assert** — это оператор: если условие ложно, программа останавливается с ошибкой. Если истинно — ничего не происходит, идём дальше.

````python
# Тест 1: обычный случай
assert calc_dti(30_000, 100_000) == 0.3

# Тест 2: DTI = 0, если нет долга
assert calc_dti(0, 100_000) == 0.0

# Тест 3: DTI может быть больше 1, если долг больше дохода
assert calc_dti(150_000, 100_000) == 1.5

print("Все тесты прошли")
````

## 3. Тест на ошибку

Хороший тест проверяет не только успешные случаи, но и **как функция падает**. Например, если `monthly_income = 0`, функция должна выбросить `ValueError`, а не поделить на ноль.

В Python для этого есть `try/except`.

````python
try:
    calc_dti(30_000, 0)
    print("ОШИБКА: функция не выбросила исключение")
except ValueError:
    print("ОК: функция корректно выбросила ValueError")
````

## 5. Мини-задача

1. Напиши функцию `calc_payment(principal, annual_rate, months)`, которая считает ежемесячный аннуитетный платёж по формуле:

```
monthly_rate = annual_rate / 12
payment = principal * monthly_rate / (1 - (1 + monthly_rate) ** (-months))
```

2. Проверь на примере: 1 000 000 ₽ под 15% годовых на 60 месяцев — платёж ≈ 23 790 ₽.

3. Напиши негативный тест: если `months = 0`, функция должна выбросить `ValueError`.

Подсказка: чтобы функция выбросила ошибку, добавь `if months <= 0: raise ValueError(...)` в начало.

````python
def calc_payment(principal, annual_rate, months):
    if months <= 0:
        raise ValueError("месяцы положительны должны быть")
    monthly_rate = annual_rate / 12
    return principal * monthly_rate / (1 - (1 + monthly_rate) ** (-months))

print(calc_payment(1000000, 0.15, 60))
````
